```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import *
```

Fashion-MNIST uses values of 0 and 255 to represent binary states. This is flattened to {0, 1}.

```python
X, y = fetch_openml("Fashion-MNIST", version=1, as_frame=False, return_X_y=True)
X = X / 255.0
y = y.astype(int)

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=10000, random_state=42, shuffle=True)
```

Set up a MLPClassifier with two hidden layers of 144 and 64 neurons, respectively. Other hyperparamaters include the ReLU acticvation function, a batch size of 96, and the default learning rate. Instead of the normal rate of learning provided, a more aggressive learning rate of 0.004 is applied.


```python
mlp = MLPClassifier(hidden_layer_sizes=(144, 64), activation='relu', solver='sgd', batch_size=96, random_state=42, learning_rate_init=0.004, early_stopping=True)
n_tr_values = [100, 500, 1000, 10000, 60000]
train_errors = []
test_errors = []


def train_model(X_tr, y_tr, X_te, y_te, n_tr_values, classifier):
    for n in n_tr_values:
        X_tr_fit = classifier.fit(X_tr[:n], y_tr[:n])
        train_errors.append(1 - accuracy_score(y_tr, X_tr_fit.predict(X_tr)))

        if n <= 10000:
            X_te_fit = classifier.fit(X_te[:n], y_te[:n])
            test_errors.append(1 - accuracy_score(y_te, X_te_fit.predict(X_te)))
        else:
            test_errors.append(0) # For graphing
    return classifier.predict(X_tr), np.asarray(train_errors), np.asarray(test_errors), classifier.validation_scores_
 
predictions, tr, te, score = train_model(X_tr, y_tr, X_te, y_te, n_tr_values, mlp)
print(f'Training errors: {tr}, testing errors: {te}, validation_scores: {max(score)}')
```

Confusion matrix of the feedforward neural network.

```python

labels = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat', 'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']
matrix = confusion_matrix(y_tr, predictions, labels=mlp.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=matrix, display_labels=labels)
disp.plot()
plt.figsize((4, 7))
plt.show()
```
